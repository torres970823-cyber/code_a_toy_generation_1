```python
#vamos jogar par ou impar

# operadores matematico
# + soma  - subtracao  / divisao * multiplicacao = recebe == igual  % modulo(resto da divisao)

# para condicionar alguma coisa:
# comando de condição (if) if quer diser (se)
# condicao da compação
# resposta
#o que fazer se nao conseguir a resposta

# exempl,o qual é o maior numero

numero1 = int(input("informe um numero "))
numero2 = int(input("informe outro numero "))

#laço de condiçao
#se o numero1 for maior que o numero2 faça(:)
if numero1 > numero2:
  print("o numero 1 é maior do que o numero 2")
elif numero1 == numero2:
  print("os valores sao iguais")
else:
  print("o numero 2 é maior do que o numero 1")
```

```python
#desafio  - criar um jogo de par ou impar

# 1 - pedor para o usuario dois numeros
# 2 - somar esses dois numeros
# 3 - descobrir se essa soma é par ou impar
# ( o desafio esta aqui)
# 4 - informar se deu par ou impar (usar if)


num1 = int(input("informe o numero 1 "))
num2 = int(input("informe o numero 2 " ))
soma = num1 + num2
print(soma)

resultado = soma % 2

if resultado == 0:
  print("o numero é par")
else:
  print("o numero é impar")

```
